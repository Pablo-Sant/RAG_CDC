#### Gerando embeddings

```python
import chromadb
```

Cria cliente e coleção

```python
db = chromadb.PersistentClient('./chroma.db')
chroma_collection = db.get_or_create_collection("documentos_llm")
```

Define o modelo de embedding no llama-index

```python
from llama_index.core import Settings
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
```

```python
Settings.embed_model = HuggingFaceEmbedding(model_name='intfloat/multilingual-e5-large')
```

Salvando Embedding no BD Chroma

```python
from llama_index.vector_stores.chroma import ChromaVectorStore
from llama_index.core import StorageContext, VectorStoreIndex
```

```python
vector_store = ChromaVectorStore(chroma_collection)
storage_context = StorageContext.from_defaults(vector_store=vector_store)
```

```python
index = VectorStoreIndex(
    nodes,
    storage_context=storage_context,
    show_progress=True
)
```

Interface gráfica com Gradio

```python
import gradio as gr
```

```python
def chat_bot(message, chat_history):
    response = chat_engine.chat(message)
    
    if chat_history is None:
        chat_history = []
        
    chat_history.append({'role':'user', 'content': message})
    chat_history.append({'role':'assistant', 'content': str(response)})
    
        
    return "", chat_history



def reset_chat():
    chat_engine.reset()
    
    return [] 
```

```python
with gr.Blocks() as demo:
    gr.Markdown("# ChatBot do Código de Defesa do Consumidor")
    
    chatbot = gr.Chatbot(label='conversa')
    msg = gr.Textbox(label='Digite a sua mensagem')
    resetar = gr.Button('Limpar')
    
    msg.submit(chat_bot, [msg, chatbot], [msg, chatbot])
    resetar.click(reset_chat, None, chatbot, queue=False)
    
demo.launch(debug=True)
```
